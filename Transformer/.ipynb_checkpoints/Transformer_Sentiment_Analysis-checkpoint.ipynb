# Movie Review Sentiment Classification with a Transformer
**A Transformer Encoder built from scratch using TensorFlow and Keras**

## Project Overview
Build a binary sentiment classifier that predicts whether an IMDB movie review is **Negative (0)** or **Positive (1)**. We implement token embeddings, positional embeddings, multi-head self-attention, a feed-forward network, residual connections, layer normalization, and padding-aware pooling. This is a compact educational Transformer Encoder, not a pretrained BERT model.

**Workflow:** Load data → Explore → Pad sequences → Build Transformer → Train → Evaluate → Predict custom reviews.

> The Keras IMDB dataset includes labels for both its training and test portions. We create a validation split from the training portion and reserve the official test set for final evaluation.

## 1. Import Libraries and Set Random Seeds
Import tools for data preparation, visualization, neural-network construction, and evaluation. Setting random seeds improves experiment reproducibility, although exact results can still vary across hardware and software versions.

```python
import os, re, random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)
sns.set_theme(style="whitegrid")

print("TensorFlow:", tf.__version__)
print("Keras:", keras.__version__)
```

## 2. Configure the Experiment
We limit vocabulary size and sequence length to keep training manageable. `EMBED_DIM` controls token-vector size, `NUM_HEADS` controls the number of attention heads, and `FF_DIM` controls the hidden width of the encoder feed-forward network.

```python
VOCAB_SIZE = 10000
MAX_LEN = 200
EMBED_DIM = 64
NUM_HEADS = 4
FF_DIM = 128
DROPOUT_RATE = 0.2
BATCH_SIZE = 64
EPOCHS = 8

print("Vocabulary:", VOCAB_SIZE, "| Maximum sequence length:", MAX_LEN)
```

## 3. Load the IMDB Dataset
Keras provides 50,000 labeled movie reviews: 25,000 for training and 25,000 for testing. Each review is already represented as integer token IDs. The label `0` means negative and `1` means positive.

```python
(x_train_raw, y_train_raw), (x_test_raw, y_test) = imdb.load_data(num_words=VOCAB_SIZE)

print("Training reviews:", len(x_train_raw))
print("Test reviews:", len(x_test_raw))
print("Training labels:", np.unique(y_train_raw, return_counts=True))
print("Test labels:", np.unique(y_test, return_counts=True))
```

## 4. Explore Review Lengths
Reviews contain different numbers of tokens. We inspect their lengths to understand why the sequences must be padded or truncated before they can be combined into batches.

```python
train_lengths = [len(review) for review in x_train_raw]
print("Shortest:", min(train_lengths))
print("Longest:", max(train_lengths))
print("Average:", round(np.mean(train_lengths), 2))
print("Median:", int(np.median(train_lengths)))

plt.figure(figsize=(9, 4))
plt.hist(train_lengths, bins=50, edgecolor="black")
plt.axvline(MAX_LEN, color="red", linestyle="--", label=f"MAX_LEN={MAX_LEN}")
plt.title("Distribution of IMDB Review Lengths")
plt.xlabel("Tokens")
plt.ylabel("Number of reviews")
plt.legend()
plt.tight_layout()
plt.show()
```

## 5. Inspect Sentiment Balance
We count negative and positive labels. This helps us understand whether one class dominates the training data.

```python
label_counts = pd.Series(y_train_raw).value_counts().sort_index()
label_counts.index = ["Negative (0)", "Positive (1)"]
print(label_counts)

plt.figure(figsize=(6, 4))
sns.barplot(x=label_counts.index, y=label_counts.values)
plt.title("Training Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Reviews")
plt.tight_layout()
plt.show()
```

## 6. Decode a Review
The integer sequences can be converted back into readable text using the IMDB word-index dictionary. The dataset reserves special IDs for start, unknown, and padding tokens.

```python
word_index = imdb.get_word_index()
reverse_word_index = {index + 3: word for word, index in word_index.items()}
reverse_word_index.update({0: "<PAD>", 1: "<START>", 2: "<OOV>", 3: "<UNUSED>"})

def decode_review(sequence):
    return " ".join(reverse_word_index.get(token, "?") for token in sequence)

print("Label:", "Positive" if y_train_raw[0] == 1 else "Negative")
print(decode_review(x_train_raw[0])[:1200])
```

## 7. Pad and Truncate Sequences
All reviews must have the same length. Short reviews are padded with `0`; long reviews are truncated. We pad at the end and retain the beginning of each review.

```python
X_train_all = pad_sequences(
    x_train_raw, maxlen=MAX_LEN, padding="post",
    truncating="post", value=0
)
X_test = pad_sequences(
    x_test_raw, maxlen=MAX_LEN, padding="post",
    truncating="post", value=0
)
y_train_all = np.asarray(y_train_raw, dtype="float32")
y_test = np.asarray(y_test, dtype="float32")

print("Training:", X_train_all.shape)
print("Test:", X_test.shape)
```

## 8. Split Training and Validation Data
We reserve 20% of the labeled training portion for validation. Stratification preserves the class balance. The official test set is not used during training or model selection.

```python
X_train, X_val, y_train, y_val = train_test_split(
    X_train_all, y_train_all,
    test_size=0.20,
    random_state=SEED,
    stratify=y_train_all
)

print("Training:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Test:", X_test.shape, y_test.shape)
```

## 9. Inspect Token IDs and Padding
Token ID `0` represents padding. Later, the Transformer will use these IDs to build an attention mask, and the pooling layer will ignore padded positions.

```python
print("Example token sequence:", X_train[0][:40])
print("Non-padding tokens:", np.count_nonzero(X_train[0]))
print("Padding tokens:", np.sum(X_train[0] == 0))
```

## 10. Implement Positional Embeddings
Self-attention does not automatically encode token order. This custom layer adds a trainable position vector to each trainable token embedding. `mask_zero=True` marks padding tokens.

```python
@keras.utils.register_keras_serializable(package="SentimentTransformer")
class PositionalEmbedding(layers.Layer):
    def __init__(self, vocab_size, max_length, embed_dim, **kwargs):
        super().__init__(**kwargs)
        self.vocab_size = vocab_size
        self.max_length = max_length
        self.embed_dim = embed_dim
        self.token_embedding = layers.Embedding(vocab_size, embed_dim, mask_zero=True)
        self.position_embedding = layers.Embedding(max_length, embed_dim)

    def call(self, token_ids):
        positions = tf.range(tf.shape(token_ids)[-1])
        token_vectors = self.token_embedding(token_ids)
        position_vectors = self.position_embedding(positions)
        return token_vectors + position_vectors

    def get_config(self):
        config = super().get_config()
        config.update({
            "vocab_size": self.vocab_size,
            "max_length": self.max_length,
            "embed_dim": self.embed_dim
        })
        return config
```

## 11. Implement a Transformer Encoder Block
The encoder block contains multi-head self-attention, a feed-forward network, residual connections, layer normalization, and dropout. The attention mask prevents padding tokens from being used as keys/values.

```python
@keras.utils.register_keras_serializable(package="SentimentTransformer")
class TransformerEncoderBlock(layers.Layer):
    def __init__(self, embed_dim, num_heads, ff_dim, dropout_rate=0.1, **kwargs):
        super().__init__(**kwargs)
        if embed_dim % num_heads != 0:
            raise ValueError("embed_dim must be divisible by num_heads.")
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.ff_dim = ff_dim
        self.dropout_rate = dropout_rate

        self.attention = layers.MultiHeadAttention(
            num_heads=num_heads,
            key_dim=embed_dim // num_heads,
            dropout=dropout_rate
        )
        self.ffn = keras.Sequential([
            layers.Dense(ff_dim, activation="relu"),
            layers.Dense(embed_dim)
        ])
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.dropout1 = layers.Dropout(dropout_rate)
        self.dropout2 = layers.Dropout(dropout_rate)

    def call(self, inputs, training=None):
        token_vectors, token_ids = inputs
        token_mask = tf.not_equal(token_ids, 0)
        attention_mask = token_mask[:, tf.newaxis, :]

        attention_output = self.attention(
            query=token_vectors,
            value=token_vectors,
            key=token_vectors,
            attention_mask=attention_mask,
            training=training
        )
        attention_output = self.dropout1(attention_output, training=training)
        out1 = self.norm1(token_vectors + attention_output)

        ffn_output = self.ffn(out1, training=training)
        ffn_output = self.dropout2(ffn_output, training=training)
        return self.norm2(out1 + ffn_output)

    def get_config(self):
        config = super().get_config()
        config.update({
            "embed_dim": self.embed_dim,
            "num_heads": self.num_heads,
            "ff_dim": self.ff_dim,
            "dropout_rate": self.dropout_rate
        })
        return config
```

## 12. Implement Padding-Aware Pooling
The encoder returns one vector per token. We need one vector for the entire review. This custom average-pooling layer averages only real tokens and excludes padding positions.

```python
@keras.utils.register_keras_serializable(package="SentimentTransformer")
class MaskedGlobalAveragePooling1D(layers.Layer):
    def call(self, inputs):
        token_vectors, token_ids = inputs
        mask = tf.cast(tf.not_equal(token_ids, 0), token_vectors.dtype)
        mask = tf.expand_dims(mask, axis=-1)
        summed = tf.reduce_sum(token_vectors * mask, axis=1)
        counts = tf.reduce_sum(mask, axis=1)
        return summed / tf.maximum(counts, 1.0)
```

## 13. Build the Transformer Classifier
We connect the token input, positional embedding, encoder block, masked pooling, and dense classification head. A sigmoid output gives the probability that a review is positive.

```python
token_inputs = keras.Input(shape=(MAX_LEN,), dtype="int32", name="token_ids")

x = PositionalEmbedding(
    VOCAB_SIZE, MAX_LEN, EMBED_DIM, name="positional_embedding"
)(token_inputs)

x = TransformerEncoderBlock(
    EMBED_DIM, NUM_HEADS, FF_DIM, DROPOUT_RATE,
    name="transformer_encoder"
)([x, token_inputs])

x = MaskedGlobalAveragePooling1D(name="masked_average_pooling")(
    [x, token_inputs]
)
x = layers.Dropout(DROPOUT_RATE)(x)
x = layers.Dense(64, activation="relu")(x)
x = layers.Dropout(DROPOUT_RATE)(x)
outputs = layers.Dense(1, activation="sigmoid", name="sentiment")(x)

model = keras.Model(token_inputs, outputs, name="IMDB_Transformer_Classifier")
model.summary()
```

## 14. Compile the Model
We use Adam optimization and binary cross-entropy because this is a two-class problem with a sigmoid output. Accuracy, AUC, precision, and recall provide complementary views of model performance.

```python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        keras.metrics.AUC(name="auc"),
        keras.metrics.Precision(name="precision"),
        keras.metrics.Recall(name="recall")
    ]
)
```

## 15. Configure Training Callbacks
Early stopping restores the best validation-loss weights. Model checkpointing saves the best model, and ReduceLROnPlateau lowers the learning rate when validation loss stops improving.

```python
os.makedirs("artifacts", exist_ok=True)

callbacks = [
    keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=2,
        restore_best_weights=True, verbose=1
    ),
    keras.callbacks.ModelCheckpoint(
        "artifacts/best_transformer_sentiment.keras",
        monitor="val_loss", save_best_only=True, verbose=1
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss", factor=0.5, patience=1,
        min_lr=1e-5, verbose=1
    )
]
```

## 16. Train the Transformer
The model learns from training reviews and is evaluated on the validation set after every epoch. Training time depends on your hardware. If it is too slow, reduce `MAX_LEN`, `BATCH_SIZE`, or `EPOCHS`.

```python
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=callbacks,
    verbose=1
)
```

## 17. Review Training History
The History object stores loss and metrics for each epoch. Displaying it helps us inspect how the model changed during training.

```python
history_df = pd.DataFrame(history.history)
display(history_df)
print("Epochs completed:", len(history.history["loss"]))
```

## 18. Plot Training and Validation Accuracy
Compare training and validation accuracy to see whether the model learns patterns that generalize to unseen reviews.

```python
plt.figure(figsize=(9, 5))
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("Transformer: Training vs Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 19. Plot Training and Validation Loss
If training loss keeps falling while validation loss rises, the model may be overfitting. These curves help diagnose that behavior.

```python
plt.figure(figsize=(9, 5))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("Transformer: Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Binary Cross-Entropy")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 20. Evaluate on the Held-Out Test Set
The official Keras IMDB test split has labels. We have not used these examples for training or validation, so this is the final evaluation.

```python
test_results = model.evaluate(X_test, y_test, batch_size=BATCH_SIZE, verbose=1)
for metric, value in zip(model.metrics_names, test_results):
    print(f"{metric}: {value:.4f}")
```

## 21. Generate Test Predictions
The sigmoid output is the positive-class probability. We use a threshold of 0.5 to convert probabilities into negative or positive class predictions.

```python
y_test_prob = model.predict(X_test, batch_size=BATCH_SIZE, verbose=1).ravel()
y_test_pred = (y_test_prob >= 0.5).astype("int32")

print("Predictions:", y_test_pred[:20])
print("Actual:     ", y_test[:20].astype("int32"))
```

## 22. Classification Report
Precision, recall, F1-score, and support show how well the model performs for negative and positive reviews.

```python
print(classification_report(
    y_test.astype("int32"),
    y_test_pred,
    target_names=["Negative", "Positive"],
    digits=4
))
```

## 23. Confusion Matrix
The diagonal contains correct predictions. Off-diagonal values show reviews whose sentiment was misclassified.

```python
cm = confusion_matrix(y_test.astype("int32"), y_test_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm, annot=True, fmt="d", cmap="Blues",
    xticklabels=["Negative", "Positive"],
    yticklabels=["Negative", "Positive"]
)
plt.title("Transformer Confusion Matrix")
plt.xlabel("Predicted Sentiment")
plt.ylabel("Actual Sentiment")
plt.tight_layout()
plt.show()
```

## 24. Inspect Individual Predictions
We inspect a few correctly and incorrectly classified reviews. This helps us understand the model's behavior beyond a single accuracy score.

```python
def show_prediction_examples(true_labels, predicted_labels, probabilities, count=6):
    true_labels = true_labels.astype("int32")
    correct = np.where(true_labels == predicted_labels)[0]
    wrong = np.where(true_labels != predicted_labels)[0]
    selected = list(correct[:count // 2]) + list(wrong[:count - count // 2])

    for idx in selected:
        actual = "Positive" if true_labels[idx] == 1 else "Negative"
        predicted = "Positive" if predicted_labels[idx] == 1 else "Negative"
        print("-" * 80)
        print(f"Actual: {actual} | Predicted: {predicted} | P(positive): {probabilities[idx]:.3f}")
        print(decode_review(x_test_raw[idx])[:700])

show_prediction_examples(y_test, y_test_pred, y_test_prob)
```

## 25. Encode a New User Review
The IMDB dataset uses a fixed vocabulary and special token IDs. This function tokenizes a new sentence, maps known words to their dataset IDs, maps unknown words to the out-of-vocabulary ID, adds the start token, and pads the result.

```python
def encode_custom_review(text, max_length=MAX_LEN):
    tokens = re.findall(r"[a-zA-Z']+", text.lower())
    sequence = [1]  # IMDB start token

    for token in tokens:
        word_id = word_index.get(token)
        sequence.append(word_id + 3 if word_id is not None else 2)

    return pad_sequences(
        [sequence], maxlen=max_length,
        padding="post", truncating="post", value=0
    )
```

## 26. Predict Custom Reviews
Try your own movie reviews. The output is a probability estimate, not a guarantee of how every person would interpret the review.

```python
def predict_sentiment(review):
    encoded = encode_custom_review(review)
    probability = float(model.predict(encoded, verbose=0)[0][0])
    sentiment = "Positive 😊" if probability >= 0.5 else "Negative 😞"

    print("Review:", review)
    print("Predicted sentiment:", sentiment)
    print(f"Positive probability: {probability:.4f}")

predict_sentiment("The movie was wonderful, emotional, and beautifully acted.")
predict_sentiment("I found this film boring and a complete waste of time.")
```

## 27. Save the Trained Model
Save the model in Keras format so it can be reused. The custom layer classes are registered for Keras serialization; keep their definitions available when loading the model in a new session.

```python
os.makedirs("artifacts", exist_ok=True)
model.save("artifacts/transformer_sentiment_classifier.keras")
print("Model saved successfully.")
```

## 28. Load and Verify the Saved Model
Reload the saved model and make one prediction to verify that the saved artifact can be used.

```python
loaded_model = keras.models.load_model(
    "artifacts/transformer_sentiment_classifier.keras"
)

sample = encode_custom_review("A moving story with excellent performances.")
probability = float(loaded_model.predict(sample, verbose=0)[0][0])
print("Positive probability:", round(probability, 4))
```
